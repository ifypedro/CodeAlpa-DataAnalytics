```python

#Put everything together

#Extracting every book on the page.

import requests
from bs4 import BeautifulSoup

url = "https://books.toscrape.com/"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

books = soup.find_all("article", class_="product_pod")

for book in books:

    title = book.h3.a["title"]

    price = book.find(
        "p", class_="price_color"
    ).text

    rating = book.find(
        "p", class_="star-rating"
    )["class"][1]

    availability = book.find(
        "p", class_="instock availability"
    ).text.strip()



"""
Storing the data

Printing the information isn't enough. We need to turn it into a dataset that we can analyze later.
"""

books_data = []

for book in books:

    title = book.h3.a["title"]

    price = book.find(
        "p", class_="price_color"
    ).text

    rating = book.find(
        "p", class_="star-rating"
    )["class"][1]

    availability = book.find(
        "p", class_="instock availability"
    ).text.strip()

    books_data.append({
        "Title": title,
        "Price": price,
        "Rating": rating,
        "Availability": availability
    })

# Display all books

    #print(title)
    #print(price)
    #print(rating)
    #print(availability)
    #print("-" * 50)
    
# confirm if its saved to book_data
     #print(books_data[:2])

"""
Converting it to Pandas
(turning the list into a proper DataFrame.)
"""

import pandas as pd

books_data_df = pd.DataFrame(books_data)
```

```python
# ======= SCRAPING MULTIPLE PAGES SUMMARY (i included the books url) =========
import requests
from bs4 import BeautifulSoup as BS
import pandas as pd

books_data_multipage = []

for page in range(1, 51):

    multipage_url = f"https://books.toscrape.com/catalogue/page-{page}.html"

    page_response = requests.get(multipage_url)

    multipage_soup = BS(page_response.text, "html.parser")

    books = multipage_soup.find_all("article", class_="product_pod")

    for book in books:

        title = book.h3.a["title"]

        price = book.find(
            "p", class_="price_color"
        ).text.strip()

        rating = book.find(
            "p", class_="star-rating"
        )["class"][1]

        availability = book.find(
            "p", class_="instock availability"
        ).text.strip()

        relative_url = book.h3.a["href"]
        product_url = "https://books.toscrape.com/catalogue/"+relative_url

        books_data_multipage.append({
            "Title": title,
            "Price": price,
            "Rating": rating,
            "Availability": availability,
            "URL":  product_url
        })

#print("Total books scraped:", len(books_data_multipage))
```

```python
# ======= Create Dataframe from the newly scarped Data =========

books_data_multipage_df = pd.DataFrame(books_data_multipage)

# Displaying The Dataframe (top 5)

books_data_multipage_df.head()

# Displaying The Dataframe (last 5)
books_data_multipage_df.tail()
```

```python
# ======= CLEANING THE DATA =========

# Clean the price column - remove both £ and Â characters
books_data_multipage_df["Price"] = books_data_multipage_df["Price"].str.replace("£", "", regex=False)
books_data_multipage_df["Price"] = books_data_multipage_df["Price"].str.replace("Â", "", regex=False)  # Remove the problematic Unicode character
books_data_multipage_df["Price"] = books_data_multipage_df["Price"].astype(float)

# Checking the price datatype
books_data_multipage_df["Price"].dtype
```
