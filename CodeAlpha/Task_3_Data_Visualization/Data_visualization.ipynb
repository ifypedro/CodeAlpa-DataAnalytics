```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("books_dataset.csv")
```

```python
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df["Rating_Numeric"] = df["Rating"].map(rating_map)
```

```python
# ===== Price Distribution Histogram =====

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Price",
    bins=20
)

plt.title("Distribution of Book Prices")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")

plt.tight_layout()
plt.show()
```

```python
# ===== Rating Distribution =====

plt.figure(figsize=(8, 6))

sns.countplot(
    data=df,
    x="Rating",
    order=["One", "Two", "Three", "Four", "Five"]
)

plt.title("Distribution of Book Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Books")

plt.tight_layout()
plt.show()
```

```python
# ===== Top 10 Most Expensive Books =====

top_expensive = df.nlargest(10, "Price").sort_values("Price")

plt.figure(figsize=(10, 7))

sns.barplot(
    data=top_expensive,
    x="Price",
    y="Title"
)

plt.title("Top 10 Most Expensive Books")
plt.xlabel("Price (£)")
plt.ylabel("Book Title")

plt.tight_layout()
plt.show()
```

```python
# ===== Top 10 Most Cheapest Books =====

top_cheap = df.nsmallest(10, "Price").sort_values("Price")

plt.figure(figsize=(25, 7))

sns.barplot(
    data=top_cheap,
    x="Price",
    y="Title"
)

plt.title("Top 10 Cheapest Books")
plt.xlabel("Price (£)")
plt.ylabel("Book Title")

plt.tight_layout()
plt.show()
```

```python
# ===== TRating vs Price =====

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Rating_Numeric",
    y="Price"
)

plt.title("Relationship Between Book Rating and Price")
plt.xlabel("Rating")
plt.ylabel("Price (£)")

plt.tight_layout()
plt.show()
```

```python
# ===== Price Outliers =====

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["Price"]
)

plt.title("Book Price Distribution and Potential Outliers")
plt.xlabel("Price (£)")

plt.tight_layout()
plt.show()
```
